```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

pd.set_option("display.max_columns", None)
```

```python
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("\nAttrition values:\n", df["Attrition"].value_counts(dropna=False))

print("\n--- Categorical columns ---")
for col in ["BusinessTravel", "Department", "EducationField",
            "Gender", "JobRole", "MaritalStatus", "OverTime"]:
    print(col, "->", df[col].unique())

print("\n--- Ordinal range checks ---")
scales = {"Education": range(1,6), "EnvironmentSatisfaction": range(1,5),
          "JobSatisfaction": range(1,5), "JobInvolvement": range(1,5),
          "RelationshipSatisfaction": range(1,5), "WorkLifeBalance": range(1,5)}
for col, valid in scales.items():
    print(col, "out-of-range:", (~df[col].isin(valid)).sum())

print("\n--- Logical consistency ---")
print("YearsAtCompany > TotalWorkingYears:", (df.YearsAtCompany > df.TotalWorkingYears).sum())
print("YearsInCurrentRole > YearsAtCompany:", (df.YearsInCurrentRole > df.YearsAtCompany).sum())
print("YearsWiDthCurrManager > YearsAtCompany:", (df.YearsWithCurrManager > df.YearsAtCompany).sum())
print("Age - TotalWorkingYears < 14:", ((df.Age - df.TotalWorkingYears) < 14).sum())
```

```python
rows_before = len(df)

# Strip whitespace in text columns
for c in df.select_dtypes(include="object").columns:
    df[c] = df[c].str.strip()

# Drop duplicates and rows missing key fields
df = df.drop_duplicates()
df = df.dropna(subset=["Attrition", "Department", "JobRole"])

# Readable labels for charts and the dashboard
df["AttritionLabel"] = df["Attrition"].map({0: "Retained", 1: "Exited"})
df["EducationLabel"] = df["Education"].map(
    {1: "Below College", 2: "College", 3: "Bachelor", 4: "Master", 5: "Doctor"})

print("Rows before:", rows_before, "| after:", len(df))
```

```python
df["BusinessTravel"] = df["BusinessTravel"].replace({
    "Travel_Rarely": "Travel Rarely",
    "Travel_Frequently": "Travel Frequently"})

df["Department"] = df["Department"].replace({
    "Research & Development": "R&D",
    "Human Resources": "HR"})

print(df["BusinessTravel"].unique())
print(df["Department"].unique())
```

```python
from statsmodels.stats.proportion import proportion_confint

total    = len(df)
exited   = int(df["Attrition"].sum())
retained = total - exited
rate     = exited / total * 100

low, high = proportion_confint(exited, total, method="wilson")

print(f"Total employees : {total}")
print(f"Retained        : {retained} ({retained/total*100:.2f}%)")
print(f"Exited          : {exited} ({rate:.2f}%)")
print(f"Overall attrition rate: {rate:.2f}%")
print(f"95% confidence interval: {low*100:.2f}% to {high*100:.2f}%")
print(f"Retained per 1 exit: {retained/exited:.1f}")
```

```python
def attr_table(col, order=None):
    t = (df.groupby(col, observed=True)
           .agg(Employees=("Attrition", "size"), Exited=("Attrition", "sum"))
           .assign(AttritionRate=lambda x: (x.Exited / x.Employees * 100).round(2))
           .reset_index())
    t["vs_Overall_pp"] = (t.AttritionRate - rate).round(2)
    if order:
        t[col] = pd.Categorical(t[col], categories=order, ordered=True)
        t = t.sort_values(col)
    else:
        t = t.sort_values("AttritionRate", ascending=False)
    return t.reset_index(drop=True)

def rate_chart(t, col, title):
    fig = px.bar(t, x=col, y="AttritionRate", text="AttritionRate",
                 color="AttritionRate", color_continuous_scale="Reds", title=title)
    fig.add_hline(y=rate, line_dash="dash", line_color="black",
                  annotation_text=f"Overall {rate:.1f}%")
    fig.update_traces(texttemplate="%{text:.1f}%", textposition="outside")
    fig.show()
```

```python
df["AgeGroup"] = pd.cut(df["Age"], bins=[17, 25, 35, 45, 55, 100],
                        labels=["18-25", "26-35", "36-45", "46-55", "56+"])

age = attr_table("AgeGroup", order=["18-25", "26-35", "36-45", "46-55", "56+"])
display(age)
rate_chart(age, "AgeGroup", "Attrition Rate by Age Group (%)")
```

```python
gender = attr_table("Gender")
display(gender)
rate_chart(gender, "Gender", "Attrition Rate by Gender (%)")

marital = attr_table("MaritalStatus")
display(marital)
rate_chart(marital, "MaritalStatus", "Attrition Rate by Marital Status (%)")
```

```python
edu_order = ["Below College", "College", "Bachelor", "Master", "Doctor"]
edu = attr_table("EducationLabel", order=edu_order)
display(edu)
rate_chart(edu, "EducationLabel", "Attrition Rate by Education Level (%)")

field = attr_table("EducationField")
display(field)
rate_chart(field, "EducationField", "Attrition Rate by Education Field (%)")
```

```python
heat = (df.pivot_table(index="AgeGroup", columns="MaritalStatus",
                       values="Attrition", aggfunc="mean", observed=True) * 100).round(1)

fig = px.imshow(heat, text_auto=True, color_continuous_scale="Reds", aspect="auto",
                title="Attrition Rate (%) by Age Group and Marital Status")
fig.show()
```

```python
from scipy.stats import chi2_contingency

for col in ["AgeGroup", "Gender", "MaritalStatus", "EducationLabel", "EducationField"]:
    chi2, p, _, _ = chi2_contingency(pd.crosstab(df[col], df["Attrition"]))
    verdict = "significant" if p < 0.05 else "not significant"
    print(f"{col:15s} chi2={chi2:6.2f}  p={p:.4f}  -> {verdict}")
```

```python
df["TenureBand"] = pd.cut(df["YearsAtCompany"],
                          bins=[-1, 1, 2, 5, 10, 20, 100],
                          labels=["0-1", "2", "3-5", "6-10", "11-20", "21+"])

tenure_order = ["0-1", "2", "3-5", "6-10", "11-20", "21+"]
tenure = attr_table("TenureBand", order=tenure_order)
display(tenure)
rate_chart(tenure, "TenureBand", "Attrition Rate by Tenure Band (years at company)")
```

```python
df["CareerStage"] = pd.cut(df["TotalWorkingYears"],
                           bins=[-1, 5, 15, 100],
                           labels=["Early (0-5 yrs)", "Mid (6-15 yrs)", "Senior (16+ yrs)"])

stage_order = ["Early (0-5 yrs)", "Mid (6-15 yrs)", "Senior (16+ yrs)"]
stage = attr_table("CareerStage", order=stage_order)
stage["Share_of_All_Exits_%"] = (stage.Exited / exited * 100).round(2)
display(stage)
rate_chart(stage, "CareerStage", "Attrition Rate by Career Stage (%)")

# Job level as a second view of seniority
level = attr_table("JobLevel", order=[1, 2, 3, 4, 5])
display(level)
rate_chart(level, "JobLevel", "Attrition Rate by Job Level (%)")
```

```python
combo = (df.groupby(["OverTime", "BusinessTravel"], observed=True)
           .agg(Employees=("Attrition", "size"), Exited=("Attrition", "sum"))
           .assign(AttritionRate=lambda x: (x.Exited / x.Employees * 100).round(2))
           .reset_index())
display(combo)

heat = (df.pivot_table(index="OverTime", columns="BusinessTravel",
                       values="Attrition", aggfunc="mean") * 100).round(1)
heat = heat[travel_order]

fig = px.imshow(heat, text_auto=True, color_continuous_scale="Reds", aspect="auto",
                title="Attrition Rate (%) by Overtime and Business Travel")
fig.show()
```

```python
df["WorkloadExposure"] = (
    (df["OverTime"] == "Yes").astype(int) +
    (df["BusinessTravel"] == "Travel Frequently").astype(int)
).map({0: "Neither", 1: "One factor", 2: "Both (OT + frequent travel)"})

expo_order = ["Neither", "One factor", "Both (OT + frequent travel)"]
expo = attr_table("WorkloadExposure", order=expo_order)
expo["WAI"] = (expo.AttritionRate / rate * 100).round(0)
expo["Share_of_All_Exits_%"] = (expo.Exited / exited * 100).round(2)
display(expo)

fig = px.bar(expo, x="WorkloadExposure", y="WAI", text="WAI",
             color="WAI", color_continuous_scale="Reds",
             title="Workload Attrition Index (100 = overall baseline)")
fig.add_hline(y=100, line_dash="dash", line_color="black")
fig.show()
```
